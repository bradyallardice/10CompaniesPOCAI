```python
# Build ~1M jobs from companies found in Data/ai_development_deduplicated.csv
# Includes ALL seed UIDs + random sample of other jobs from those companies
# Exports only: uid, company_id, content_clean, duplicate_group

import os
import math
import csv
import psycopg2
import pandas as pd
from io import StringIO
from dotenv import load_dotenv

# -------------------
# Config
# -------------------
SEED_FILE = "Data/ai_development_deduplicated_custom.csv"
OUTPUT_FILE = "Data/ai_company_jobs_target1M.csv"
TARGET_TOTAL = 1_000_000  # desired total rows
UID_COL = "uid"           # must exist in seed file

# -------------------
# Load seed CSV
# -------------------
seed = pd.read_csv(SEED_FILE)

if UID_COL not in seed.columns:
    raise ValueError(f"Column '{UID_COL}' not found in {SEED_FILE}")

# Prefer company_id if present; otherwise fall back to company_name
if "company_id" in seed.columns and seed["company_id"].notna().any():
    COMPANY_MODE = "id"
    COMPANY_COL = "company_id"
    seed_companies = (
        seed[COMPANY_COL]
        .dropna()
        .astype(str)
        .drop_duplicates()
        .tolist()
    )
else:
    if "company_name" not in seed.columns:
        raise ValueError(f"Neither 'company_id' nor 'company_name' present/useful in {SEED_FILE}")
    COMPANY_MODE = "name"
    COMPANY_COL = "company_name"
    seed_companies = (
        seed[COMPANY_COL]
        .dropna()
        .astype(str)
        .drop_duplicates()
        .tolist()
    )

# UIDs as strings (must be valid UUIDs in DB)
seed_uids = (
    seed[UID_COL]
    .dropna()
    .astype(str)
    .drop_duplicates()
    .tolist()
)

print(f"Seed companies ({'company_id' if COMPANY_MODE=='id' else 'company_name'}): {len(seed_companies):,}")
print(f"Seed UIDs: {len(seed_uids):,}")

# -------------------
# Connect to DB
# -------------------
load_dotenv("config.env")
conn = psycopg2.connect(
    dbname=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST", "localhost"),
    port=os.getenv("DB_PORT", "5432"),
)
conn.autocommit = True
conn.set_client_encoding('UTF8')
cur = conn.cursor()

# -------------------
# Create TEMP tables & bulk load (CSV-safe)
# -------------------
cur.execute("DROP TABLE IF EXISTS temp_companies;")
if COMPANY_MODE == "id":
    cur.execute("CREATE TEMP TABLE temp_companies(company_id text);")
else:
    cur.execute("CREATE TEMP TABLE temp_companies(company_name text);")

cur.execute("DROP TABLE IF EXISTS temp_uids;")
# Make temp_uids UUID so we can join jp.uid = tu.uid without casting (uses index on jp.uid)
cur.execute("CREATE TEMP TABLE temp_uids(uid uuid);")

def copy_list_to_temp(cur, table, col, items):
    sio = StringIO()
    w = csv.writer(sio)  # proper CSV quoting for commas/newlines
    for x in items:
        w.writerow([("" if x is None else str(x).strip())])
    sio.seek(0)
    cur.copy_expert(f"COPY {table}({col}) FROM STDIN WITH (FORMAT csv)", sio)

# Load companies/uids
if COMPANY_MODE == "id":
    copy_list_to_temp(cur, "temp_companies", "company_id", seed_companies)
    cur.execute("CREATE INDEX ON temp_companies(company_id);")
else:
    copy_list_to_temp(cur, "temp_companies", "company_name", seed_companies)
    cur.execute("CREATE INDEX ON temp_companies(company_name);")

# COPY will auto-cast valid UUID strings into uuid column
copy_list_to_temp(cur, "temp_uids", "uid", seed_uids)
cur.execute("CREATE INDEX ON temp_uids(uid);")

# Sanity counts (optional)
cur.execute("SELECT COUNT(*) FROM temp_companies"); print("temp_companies rows:", cur.fetchone()[0])
cur.execute("SELECT COUNT(*) FROM temp_uids"); print("temp_uids rows:", cur.fetchone()[0])

# -------------------
# Count already-included (seed) rows
# -------------------
cur.execute("""
    SELECT COUNT(*) 
    FROM job_postings jp 
    JOIN temp_uids tu ON jp.uid = tu.uid
    WHERE jp.content_clean IS NOT NULL
""")
included_ct = cur.fetchone()[0]
print(f"Already-included rows (seed UIDs): {included_ct:,}")

need_more = max(TARGET_TOTAL - included_ct, 0)
if need_more == 0:
    print("We already meet or exceed the target; exporting seed rows only.")

# -------------------
# Count eligible extras (same companies, not in seed uids)
# -------------------
if COMPANY_MODE == "id":
    join_company = "tc.company_id = jp.company_id::text"  # temp is text; cast jp side (unknown base type)
else:
    join_company = "tc.company_name = jp.company_name"

cur.execute(f"""
    SELECT COUNT(*)
    FROM job_postings jp
    JOIN temp_companies tc ON {join_company}
    LEFT JOIN temp_uids tu ON tu.uid = jp.uid
    WHERE tu.uid IS NULL
      AND jp.content_clean IS NOT NULL
""")
eligible_ct = cur.fetchone()[0]
print(f"Eligible additional rows: {eligible_ct:,}")

# -------------------
# Decide sampling modulus k (0..1000), md5-based predicate
# -------------------
if need_more > 0 and eligible_ct > 0:
    frac = need_more / float(eligible_ct)
    k = min(1000, max(0, math.ceil(1000 * frac)))
    if k == 0:
        k = 1  # ensure some sampling; LIMIT will cap to need_more
else:
    k = 0
print(f"Sampling modulus k: {k} (out of 1000)")

# -------------------
# Seed SQL (ALL seed UIDs, exact match)
# -------------------
seed_sql = """
    SELECT jp.uid::text, jp.company_id::text, jp.content_clean, jp.duplicate_group
    FROM job_postings jp
    JOIN temp_uids tu ON jp.uid = tu.uid
    WHERE jp.content_clean IS NOT NULL
"""

# -------------------
# Extra SQL (sample from same companies, excluding seed uids)
# Deterministic md5-based sampling; LIMIT caps to need_more
# -------------------
if need_more > 0 and eligible_ct > 0 and k > 0:
    extra_sql = f"""
        SELECT jp.uid::text, jp.company_id::text, jp.content_clean, jp.duplicate_group
        FROM job_postings jp
        JOIN temp_companies tc ON {join_company}
        LEFT JOIN temp_uids tu ON tu.uid = jp.uid
        WHERE tu.uid IS NULL
          AND jp.content_clean IS NOT NULL
          AND ( ('x' || substr(md5(jp.uid::text), 1, 8))::bit(32)::int % 1000 ) < {k}
        LIMIT {need_more}
    """
else:
    # empty result
    extra_sql = """
        SELECT uid::text, company_id::text, content_clean, duplicate_group
        FROM (SELECT NULL::text uid, NULL::text company_id, NULL::text content_clean, NULL::text duplicate_group) x
        WHERE false
    """

# -------------------
# UNION ALL + DISTINCT ON (uid) so seed rows win
# -------------------
final_sql = f"""
WITH seed AS (
  {seed_sql}
),
extra AS (
  {extra_sql}
),
unioned AS (
  SELECT 1 AS src, * FROM seed
  UNION ALL
  SELECT 2 AS src, * FROM extra
)
SELECT uid, company_id, content_clean, duplicate_group
FROM (
  SELECT DISTINCT ON (uid) uid, company_id, content_clean, duplicate_group, src
  FROM unioned
  ORDER BY uid, src        -- src=1 (seed) preferred over src=2 (extra)
) t
"""

print("Exporting to CSV via COPY ...")
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    cur.copy_expert(
        f"COPY ({final_sql}) TO STDOUT WITH CSV HEADER",
        f
    )

# -------------------
# Sanity checks
# -------------------
out_df = pd.read_csv(OUTPUT_FILE, usecols=["uid", "company_id", "content_clean", "duplicate_group"])
print(f"Wrote {len(out_df):,} rows to {OUTPUT_FILE}")
assert out_df["uid"].is_unique, "Duplicate UIDs detected in output (should be unique)"
print("UID uniqueness check passed ✅")

cur.close()
conn.close()
print("Done.")
```

```python
import gzip, json, io
from collections import Counter

# If you have ijson, we'll use it for streaming arrays
try:
    import ijson
    HAVE_IJSON = True
except Exception:
    HAVE_IJSON = False

path = "/Users/bradyallardice/Downloads/240722_panel_data_data_000000000002.json.gzip"

# Try these field names; add more if needed.
CANDIDATES = ["duplicate_group", "duplicateGroup", "dup_group", "duplicate_group_id"]

def sniff_format_gz(path) -> str:
    """Return 'ndjson' or 'array' by peeking the first non-whitespace char."""
    with gzip.open(path, "rt", encoding="utf-8", errors="ignore") as f:
        while True:
            ch = f.read(1)
            if not ch:
                return "ndjson"   # empty -> treat as ndjson
            if ch.isspace():
                continue
            return "array" if ch == "[" else "ndjson"

def pick_field(record):
    for k in CANDIDATES:
        if k in record:
            return k
    return None

def normalize_group(val):
    """Return a normalized string id or None for null/empty."""
    if val is None:
        return None
    # sometimes the field can be a list; stringify consistently
    if isinstance(val, (list, tuple)):
        if not val:
            return None
        val = "|".join(str(x) for x in val)
    s = str(val).strip()
    return s if s else None

fmt = sniff_format_gz(path)
print(f"Detected JSON format: {fmt.upper()}")

total = 0
nulls = 0
group_counts = Counter()
chosen_col = None

if fmt == "ndjson":
    with gzip.open(path, "rt", encoding="utf-8", errors="ignore") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                rec = json.loads(line)
            except json.JSONDecodeError:
                continue
            if chosen_col is None:
                chosen_col = pick_field(rec)
                if chosen_col is None:
                    raise RuntimeError(f"None of {CANDIDATES} present in records.")
            total += 1
            g = normalize_group(rec.get(chosen_col))
            if g is None:
                nulls += 1
            else:
                group_counts[g] += 1

elif fmt == "array":
    if HAVE_IJSON:
        with gzip.open(path, "rb") as f:
            for rec in ijson.items(f, "item"):
                if chosen_col is None:
                    chosen_col = pick_field(rec)
                    if chosen_col is None:
                        raise RuntimeError(f"None of {CANDIDATES} present in records.")
                total += 1
                g = normalize_group(rec.get(chosen_col))
                if g is None:
                    nulls += 1
                else:
                    group_counts[g] += 1
    else:
        # Fallback: one-shot load (may be memory heavy)
        import pandas as pd
        print("⚠️ ijson not installed; falling back to full load with pandas (no chunksize).")
        df = pd.read_json(path, lines=False, compression="gzip")
        # choose column
        for c in CANDIDATES:
            if c in df.columns:
                chosen_col = c
                break
        if chosen_col is None:
            raise RuntimeError(f"None of {CANDIDATES} found. Columns: {list(df.columns)[:50]}")
        s = df[chosen_col]
        total = len(s)
        # treat empty strings as null
        null_mask = s.isna() | s.astype(str).str.strip().eq("")
        nulls = int(null_mask.sum())
        # count groups (normalize lists)
        def norm(x):
            if isinstance(x, (list, tuple)):
                if not x: return None
                x = "|".join(str(v) for v in x)
            x = None if x is None else str(x).strip() or None
            return x
        vals = [norm(v) for v in s.tolist()]
        for v in vals:
            if v is not None:
                group_counts[v] += 1

nonnull_rows = total - nulls
n_groups = len(group_counts)
groups_size1 = sum(1 for n in group_counts.values() if n == 1)
groups_size_gt1 = sum(1 for n in group_counts.values() if n > 1)
rows_in_groups_gt1 = sum(n for n in group_counts.values() if n > 1)

print("\n=== Duplicate Group Coverage (from gzip JSON) ===")
print(f"Field used: {chosen_col}")
print(f"Total rows: {total:,}")
print(f"Rows with {chosen_col}=NULL/empty: {nulls:,} ({nulls/total:.1%})")
print(f"Rows with {chosen_col} present:   {nonnull_rows:,} ({nonnull_rows/total:.1%})")
print(f"Distinct {chosen_col} values:     {n_groups:,}")
print(f"  groups with size = 1:           {groups_size1:,}")
print(f"  groups with size > 1:           {groups_size_gt1:,}")
print(f"Rows in size>1 groups:            {rows_in_groups_gt1:,} "
      f"({rows_in_groups_gt1/total:.1%} of all rows)")
```

```python
import os
import pandas as pd
import psycopg2
from dotenv import load_dotenv

# ---------- connect ----------
load_dotenv("config.env")
conn = psycopg2.connect(
    dbname=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST", "localhost"),
    port=os.getenv("DB_PORT", "5432"),
)
conn.autocommit = True

def qdf(query, params=None):
    return pd.read_sql_query(query, conn, params=params)

def Q_ident(s: str) -> str:
    """Quote an identifier as "schema" or "table" safely for plain-string SQL."""
    return '"' + s.replace('"', '""') + '"'

def QT(schema: str, table: str) -> str:
    return f"{Q_ident(schema)}.{Q_ident(table)}"

print("Connected.")

# ---------- find tables that have duplicate_group ----------
cand = qdf("""
SELECT c.table_schema, c.table_name, c.column_name, c.data_type
FROM information_schema.columns c
WHERE c.column_name = 'duplicate_group'
ORDER BY c.table_schema, c.table_name
""")

if cand.empty:
    print("No exact 'duplicate_group' column found. Looking for similar names...")
    similar = qdf("""
        SELECT c.table_schema, c.table_name, c.column_name, c.data_type
        FROM information_schema.columns c
        WHERE c.column_name ILIKE '%dup%'
        ORDER BY c.table_schema, c.table_name, c.column_name
    """)
    display(similar.head(50))
    raise SystemExit("⚠️ No exact column; check the similar list above.")

# choose the largest table that has the column
sizes = qdf("""
SELECT n.nspname AS table_schema, c.relname AS table_name, c.reltuples::bigint AS est_rows
FROM pg_class c
JOIN pg_namespace n ON n.oid = c.relnamespace
WHERE c.relkind IN ('r','p','m','v')
""")
target = (cand.merge(sizes, on=["table_schema","table_name"], how="left")
              .sort_values("est_rows", ascending=False)
              .iloc[0])
SCHEMA = target["table_schema"]
TABLE  = target["table_name"]
DTYPE  = target["data_type"]
TQUAL  = QT(SCHEMA, TABLE)
print(f"Target: {SCHEMA}.{TABLE}  (est_rows≈{int(target['est_rows'] or 0):,})  duplicate_group type: {DTYPE}")

# ---------- basic counts ----------
total = qdf(f"SELECT COUNT(*) AS n FROM {TQUAL}").iloc[0,0]
nulls = qdf(f"SELECT COUNT(*) AS n FROM {TQUAL} WHERE duplicate_group IS NULL").iloc[0,0]
print(f"\nTotal rows: {total:,}")
print(f"NULL duplicate_group: {nulls:,} ({(nulls/total*100 if total else 0):.1f}%)")

# blanks only make sense if it's text-like
if any(x in (DTYPE or "").lower() for x in ["char", "text"]):
    blanks = qdf(f"""
        SELECT COUNT(*) AS n
        FROM {TQUAL}
        WHERE duplicate_group IS NOT NULL AND btrim(duplicate_group) = ''
    """).iloc[0,0]
    print(f"Blank duplicate_group: {blanks:,} ({(blanks/total*100 if total else 0):.1f}%)")

# distinct non-null groups
distinct_non_null = qdf(f"""
    SELECT COUNT(DISTINCT duplicate_group) AS n
    FROM {TQUAL}
    WHERE duplicate_group IS NOT NULL
""").iloc[0,0]
print(f"Distinct non-null duplicate_group: {distinct_non_null:,}")

# top group sizes (confirms grouping is functioning)
top_groups = qdf(f"""
    SELECT duplicate_group, COUNT(*) AS ct
    FROM {TQUAL}
    WHERE duplicate_group IS NOT NULL
    GROUP BY duplicate_group
    ORDER BY ct DESC
    LIMIT 10
""")
print("\nTop duplicate_group sizes:")
display(top_groups)

# ---------- month-by-month NULL rate (auto-picks a timestamp) ----------
ts_cols = qdf(f"""
    SELECT column_name, data_type
    FROM information_schema.columns
    WHERE table_schema = %s AND table_name = %s
      AND data_type IN ('timestamp without time zone','timestamp with time zone','date')
""", params=(SCHEMA, TABLE))
prefer = ['tst_created','created_at','posted_at','date_posted','ingested_at','updated_at','date','timestamp']
ts_col = next((c for c in prefer if (ts_cols['column_name'] == c).any()), None)
if ts_col is None and not ts_cols.empty:
    ts_col = ts_cols.iloc[0]['column_name']

if ts_col:
    print(f"\nUsing time column for trend analysis: {ts_col}")
    trend = qdf(f"""
        SELECT
          date_trunc('month', {Q_ident(ts_col)}::timestamp) AS month,
          COUNT(*) AS total_rows,
          SUM(CASE WHEN duplicate_group IS NULL THEN 1 ELSE 0 END) AS null_rows,
          ROUND(100.0 * SUM(CASE WHEN duplicate_group IS NULL THEN 1 ELSE 0 END) / NULLIF(COUNT(*),0), 2) AS null_pct
        FROM {TQUAL}
        WHERE {Q_ident(ts_col)} IS NOT NULL
        GROUP BY 1
        ORDER BY 1
    """)
    display(trend)
else:
    print("\n(No timestamp/date column found for month-by-month breakdown.)")

# ---------- which source/batch drives NULLs (if such columns exist) ----------
possible_dims = [
    'source', 'source_name', 'source_site', 'source_domain',
    'batch_id', 'etl_run_id', 'import_id', 'file_id',
    'country', 'language', 'platform'
]
present_cols = qdf("""
SELECT column_name
FROM information_schema.columns
WHERE table_schema = %s AND table_name = %s
""", params=(SCHEMA, TABLE))['column_name'].tolist()

for dim in [c for c in possible_dims if c in present_cols]:
    print(f"\nNULL breakdown by {dim} (top 15 by null count):")
    by_dim = qdf(f"""
        SELECT {Q_ident(dim)} AS {dim},
               COUNT(*) AS total_rows,
               SUM(CASE WHEN duplicate_group IS NULL THEN 1 ELSE 0 END) AS null_rows,
               ROUND(100.0 * SUM(CASE WHEN duplicate_group IS NULL THEN 1 ELSE 0 END) / NULLIF(COUNT(*),0), 2) AS null_pct
        FROM {TQUAL}
        GROUP BY 1
        HAVING COUNT(*) > 0
        ORDER BY null_rows DESC
        LIMIT 15
    """)
    display(by_dim)

print("\n✅ Diagnostics complete.")
```

```python
import os, gzip, json, sqlite3, glob, itertools, hashlib
from datetime import datetime

# ============================
# CONFIG — EDIT THESE PATHS
# ============================
# Existing (“already downloaded”) files used to build the UID index
EXISTING_GLOB = os.path.expanduser("~/Downloads/240826_panel_data_with_dg/*.json.gzip")
# New files to check (e.g., 230907_export_export000000000003.gzip etc.)
NEW_GLOB = os.path.expanduser("~/Downloads/230907_export_export*.gzip")

# Field name candidates
UID_CANDS = ["uid", "job_uid", "id"]
DATE_CANDS = ["tst_created", "created_at", "published_at", "date"]
DUPG_CANDS = ["duplicate_group", "duplicateGroup"]

# SQLite index file (created if missing; safe to delete afterwards)
UID_DB = os.path.expanduser("~/uid_index.sqlite")

# Sampling for pretty head output
RAW_HEAD_LINES = 3     # raw text lines to show per file
PARSED_HEAD_OBJS = 3   # how many parsed objects to print per file

# ============================
# Helpers
# ============================
def detect_format(path):
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        sample = f.read(4096)
    s = sample.lstrip()
    if s.startswith('['):
        return "json_array"
    if s.startswith('{') or "\n{" in sample:
        return "ndjson"
    return "unknown"

def iter_ndjson(path):
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        for line in f:
            line = line.strip()
            if not line: 
                continue
            try:
                yield json.loads(line)
            except json.JSONDecodeError:
                continue

def iter_json_array(path):
    # lightweight streaming of a gzip'd JSON array
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        # move to '['
        ch = f.read(1)
        while ch and ch.isspace():
            ch = f.read(1)
        if ch != '[':
            return
        buf = []
        depth = 0
        in_str = False
        esc = False
        # read after '['
        while True:
            ch = f.read(1)
            if not ch:
                return
            if in_str:
                buf.append(ch)
                if esc:
                    esc = False
                elif ch == '\\':
                    esc = True
                elif ch == '"':
                    in_str = False
                continue
            if ch.isspace() or ch == ',':
                continue
            if ch == ']':
                return
            if ch == '{':
                buf = ['{']
                depth = 1
                in_str = False
                esc = False
                while True:
                    ch = f.read(1)
                    if not ch:
                        return
                    buf.append(ch)
                    if in_str:
                        if esc:
                            esc = False
                        elif ch == '\\':
                            esc = True
                        elif ch == '"':
                            in_str = False
                    else:
                        if ch == '"':
                            in_str = True
                        elif ch == '{':
                            depth += 1
                        elif ch == '}':
                            depth -= 1
                            if depth == 0:
                                try:
                                    yield json.loads("".join(buf))
                                except Exception:
                                    pass
                                break

def iter_records(path):
    fmt = detect_format(path)
    if fmt == "ndjson":
        return iter_ndjson(path)
    elif fmt == "json_array":
        return iter_json_array(path)
    else:
        # best effort: try ndjson
        return iter_ndjson(path)

def pick_field(obj, cands):
    for c in cands:
        if c in obj:
            return c
    return None

def parse_dt(s):
    if s is None:
        return None
    if not isinstance(s, str):
        s = str(s)
    try:
        return datetime.fromisoformat(s.replace('Z', '+00:00'))
    except Exception:
        return None

def print_raw_head(path, n=RAW_HEAD_LINES):
    print(f"\n--- raw head of {os.path.basename(path)} ({n} lines) ---")
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        for i, line in zip(range(1, n+1), f):
            print(f"{i:>3}: {line.rstrip()[:300]}")

# ============================
# Build UID index (SQLite)
# ============================
def make_uid_index(db_path):
    conn = sqlite3.connect(db_path)
    cur = conn.cursor()
    cur.execute("PRAGMA journal_mode=WAL;")
    cur.execute("PRAGMA synchronous=OFF;")
    cur.execute("""
        CREATE TABLE IF NOT EXISTS uids (
            uid TEXT PRIMARY KEY,
            src TEXT
        );
    """)
    cur.execute("CREATE INDEX IF NOT EXISTS idx_src ON uids(src);")
    conn.commit()
    return conn

def add_file_to_index(conn, path, uid_field=None):
    cur = conn.cursor()
    inserted = 0
    batch = []
    # print head for usability
    print_raw_head(path)
    # parse a few objects to pick fields
    test_records = []
    for i, obj in enumerate(iter_records(path)):
        test_records.append(obj)
        if i >= PARSED_HEAD_OBJS - 1:
            break
    if not test_records:
        print(f"(!) No records found in {os.path.basename(path)}")
        return 0, None, None, uid_field

    if uid_field is None:
        uid_field = pick_field(test_records[0], UID_CANDS)
    date_field = pick_field(test_records[0], DATE_CANDS)
    dup_field  = pick_field(test_records[0], DUPG_CANDS)

    print(f"Detected fields → uid: {uid_field}, date: {date_field}, duplicate_group: {dup_field}")
    print("--- parsed head ---")
    for j, t in enumerate(test_records, 1):
        print(f"{j:>3}: ", {k: t.get(k) for k in [uid_field, date_field, dup_field, "title", "company_name"] if k})

    # now stream all to index uids and compute date range + dup stats
    total = 0
    null_dup = 0
    dt_min = None
    dt_max = None

    def flush():
        nonlocal batch, inserted
        if not batch:
            return
        cur.executemany("INSERT OR IGNORE INTO uids(uid, src) VALUES (?, ?);", batch)
        conn.commit()
        inserted += cur.rowcount if cur.rowcount is not None else 0
        batch = []

    # include the head ones first
    for obj in test_records:
        total += 1
        uid = obj.get(uid_field)
        if uid:
            batch.append((str(uid), os.path.basename(path)))
        if date_field:
            dt = parse_dt(obj.get(date_field))
            if dt:
                dt_min = dt if dt_min is None else min(dt_min, dt)
                dt_max = dt if dt_max is None else max(dt_max, dt)
        if dup_field:
            dg = obj.get(dup_field)
            if dg is None or (isinstance(dg, str) and dg.strip() == ""):
                null_dup += 1
    flush()

    # continue streaming the rest (skip the number we already consumed)
    consumed = len(test_records)
    for i, obj in enumerate(itertools.islice(iter_records(path), consumed, None), start=consumed+1):
        total += 1
        if uid_field:
            uid = obj.get(uid_field)
            if uid:
                batch.append((str(uid), os.path.basename(path)))
        if date_field:
            dt = parse_dt(obj.get(date_field))
            if dt:
                dt_min = dt if dt_min is None else min(dt_min, dt)
                dt_max = dt if dt_max is None else max(dt_max, dt)
        if dup_field:
            dg = obj.get(dup_field)
            if dg is None or (isinstance(dg, str) and dg.strip() == ""):
                null_dup += 1
        if len(batch) >= 20_000:
            flush()
    flush()

    null_rate = (null_dup / total * 100) if total else 0.0
    print(f"Indexed {inserted:,} new uids from {os.path.basename(path)} | rows seen: {total:,} | "
          f"duplicate_group NULL/empty: {null_dup:,} ({null_rate:.2f}%)")
    if dt_min and dt_max:
        print(f"Date range: {dt_min} → {dt_max}")
    return inserted, dt_min, dt_max, uid_field

# ============================
# Overlap test for NEW files
# ============================
def overlap_check(conn, path, uid_field_hint=None):
    cur = conn.cursor()
    fmt = detect_format(path)
    # pick fields using first object
    it = iter_records(path)
    try:
        first = next(it)
    except StopIteration:
        print(f"(!) No records in {os.path.basename(path)}")
        return
    uid_field = uid_field_hint or pick_field(first, UID_CANDS)
    date_field = pick_field(first, DATE_CANDS)

    # print a tiny head
    print_raw_head(path)
    print("--- parsed head ---")
    print({k: first.get(k) for k in [uid_field, date_field, "duplicate_group", "title", "company_name"] if k})

    total = 0
    in_idx = 0
    not_in_idx = 0
    dt_min = parse_dt(first.get(date_field)) if date_field else None
    dt_max = dt_min

    # include the first record
    if uid_field and first.get(uid_field):
        uid = str(first.get(uid_field))
        row = cur.execute("SELECT 1 FROM uids WHERE uid = ? LIMIT 1;", (uid,)).fetchone()
        if row: in_idx += 1
        else:   not_in_idx += 1
        total += 1

    # stream the rest
    for obj in it:
        total += 1
        if uid_field and obj.get(uid_field):
            uid = str(obj.get(uid_field))
            row = cur.execute("SELECT 1 FROM uids WHERE uid = ? LIMIT 1;", (uid,)).fetchone()
            if row: in_idx += 1
            else:   not_in_idx += 1
        if date_field:
            dt = parse_dt(obj.get(date_field))
            if dt:
                dt_min = dt if dt_min is None else min(dt_min, dt)
                dt_max = dt if dt_max is None else max(dt_max, dt)

    pct_in = (in_idx/total*100) if total else 0.0
    print(f"\nOverlap for {os.path.basename(path)}")
    print(f"  total rows:     {total:,}")
    print(f"  in EXISTING:    {in_idx:,} ({pct_in:.2f}%)")
    print(f"  not in EXISTING:{not_in_idx:,} ({100-pct_in:.2f}%)")
    if dt_min and dt_max:
        print(f"  date range:     {dt_min} → {dt_max}")

# ============================
# Run: build index from EXISTING, then check NEW
# ============================
existing_files = sorted(glob.glob(EXISTING_GLOB))
new_files      = sorted(glob.glob(NEW_GLOB))

print(f"Found {len(existing_files)} EXISTING files")
print(f"Found {len(new_files)} NEW files")

if not existing_files:
    raise SystemExit("No EXISTING files matched. Edit EXISTING_GLOB.")

conn = make_uid_index(UID_DB)

# Build UID index (once). If you already created it earlier, you can skip this loop.
seen_uid_field = None
all_dt_min = None
all_dt_max = None
for fp in existing_files:
    added, dmin, dmax, seen_uid_field = add_file_to_index(conn, fp, uid_field=seen_uid_field)
    if dmin and dmax:
        all_dt_min = dmin if all_dt_min is None else min(all_dt_min, dmin)
        all_dt_max = dmax if all_dt_max is None else max(all_dt_max, dmax)

# How big is the index?
cur = conn.cursor()
count_index = cur.execute("SELECT COUNT(*) FROM uids;").fetchone()[0]
print(f"\nUID index size: {count_index:,} unique uids")
if all_dt_min and all_dt_max:
    print(f"EXISTING overall date range: {all_dt_min} → {all_dt_max}")

# Check each NEW file for overlap + its own date range & head
for fp in new_files:
    overlap_check(conn, fp, uid_field_hint=seen_uid_field)

# Close when done
conn.close()
print("\nDone.")
```

```python
import os, gzip, json, hashlib, sqlite3, re
from glob import glob
from urllib.parse import urlparse, parse_qsl, urlunparse
from datetime import datetime
from collections import defaultdict

import os, sqlite3, tempfile
from pathlib import Path

# pick a path & make sure its directory exists
DB_PATH = os.path.expanduser("~/tmp/overlap_index.sqlite")  # or change to something you like
p = Path(DB_PATH).expanduser()
p.parent.mkdir(parents=True, exist_ok=True)  # <-- create ~/tmp if missing

# if a stale file exists, remove it (optional)
if p.exists():
    try: p.unlink()
    except Exception as e: print("Could not remove old DB:", e)

# connect; if this location still fails, fall back to /tmp
try:
    con = sqlite3.connect(str(p))
except sqlite3.OperationalError as e:
    print("Primary location failed:", e, "\nFalling back to /tmp …")
    DB_PATH = os.path.join(tempfile.gettempdir(), "overlap_index.sqlite")
    con = sqlite3.connect(DB_PATH)

cur = con.cursor()
print("Using SQLite at:", DB_PATH)


# ---------- CONFIG ----------
DG_DIR = os.path.expanduser("~/Downloads/240826_panel_data_with_dg")   # folder with data_*.json.gzip
OLD_FILE = os.path.expanduser("~/Downloads/240722_panel_data_data_000000000002.json.gzip")
DB_PATH = os.path.expanduser("~/tmp/overlap_index.sqlite")             # temp local index

# ---------- Helpers: streaming JSON(.gzip) ----------
def detect_format(path):
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        sample = f.read(4096).lstrip()
    if sample.startswith('['): return "json_array"
    if sample.startswith('{') or "\n{" in sample: return "ndjson"
    return "unknown"

def iter_ndjson(path):
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        for line in f:
            line=line.strip()
            if not line: continue
            try: yield json.loads(line)
            except json.JSONDecodeError: continue

def iter_json_array(path):
    with gzip.open(path, "rt", encoding="utf-8", errors="replace") as f:
        buf=[]; in_obj=False; depth=0; in_str=False; esc=False
        # seek to '['
        while True:
            c=f.read(1); 
            if not c: return
            if c=='[': break
        while True:
            c=f.read(1)
            if not c: return
            if not in_obj:
                if c=='{':
                    in_obj=True; buf=['{']; depth=1; in_str=False; esc=False
                continue
            buf.append(c)
            if in_str:
                if esc: esc=False
                elif c=='\\': esc=True
                elif c=='"': in_str=False
            else:
                if c=='"': in_str=True
                elif c=='{': depth+=1
                elif c=='}':
                    depth-=1
                    if depth==0:
                        try: yield json.loads("".join(buf))
                        except Exception: pass
                        in_obj=False

def iter_records(path):
    fmt = detect_format(path)
    return iter_ndjson(path) if fmt == "ndjson" else iter_json_array(path)

# ---------- Key builders ----------
ID_PARAMS = {"id","jobid","job_id","vacancy_id"}  # keep these query params if present

def normalize_url(u:str)->str:
    if not u: return ""
    try:
        p = urlparse(u)
        netloc = (p.netloc or "").lower()
        path = (p.path or "").rstrip('/') or '/'
        # keep only id-like params, sorted
        qdict = [(k,v) for k,v in parse_qsl(p.query, keep_blank_values=True) if k.lower() in ID_PARAMS]
        qdict.sort()
        q = "&".join(f"{k}={v}" for k,v in qdict)
        norm = urlunparse((p.scheme.lower(), netloc, path, "", q, ""))
        return norm
    except Exception:
        return u.strip()

def url_hash(u:str)->str:
    return hashlib.sha1(normalize_url(u).encode("utf-8",errors="ignore")).hexdigest()

_ws = re.compile(r"\s+")
def soft_key(company:str, title:str, tst_created:str)->str:
    def clean(s):
        if s is None: return ""
        s = s.strip().lower()
        s = _ws.sub(" ", s)
        return s
    # use date (day) granularity
    day = ""
    if tst_created:
        try:
            day = datetime.fromisoformat(str(tst_created).replace("Z","+00:00")).date().isoformat()
        except Exception:
            day = str(tst_created)[:10]
    key = f"{clean(company)}||{clean(title)}||{day}"
    return hashlib.sha1(key.encode("utf-8")).hexdigest()

# ---------- SQLite index ----------
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)
con = sqlite3.connect(DB_PATH)
cur = con.cursor()
cur.execute("PRAGMA journal_mode=WAL;")
cur.execute("CREATE TABLE urls_dg (url_hash TEXT PRIMARY KEY, example_url TEXT)")
cur.execute("CREATE TABLE soft_dg (soft_hash TEXT PRIMARY KEY, ex_company TEXT, ex_title TEXT, ex_day TEXT)")
cur.execute("CREATE INDEX IF NOT EXISTS idx_soft ON soft_dg(soft_hash)")
con.commit()

# ---------- Pass 1: index the with-dg bundle ----------
dg_files = sorted(glob(os.path.join(DG_DIR, "*.json.gzip")))
if not dg_files:
    raise SystemExit(f"No files found in {DG_DIR}")

dg_min_dt = None
dg_max_dt = None
urls_seen = 0
soft_seen = 0

def up_minmax(dt_str, cur_min, cur_max):
    try:
        dt = datetime.fromisoformat(str(dt_str).replace("Z","+00:00"))
        cur_min = dt if cur_min is None or dt < cur_min else cur_min
        cur_max = dt if cur_max is None or dt > cur_max else cur_max
    except Exception:
        pass
    return cur_min, cur_max

BATCH = 5000
buf_url = []
buf_soft = []

for path in dg_files:
    for rec in iter_records(path):
        u = rec.get("url")
        t = rec.get("title")
        c = rec.get("company_name")
        tc = rec.get("tst_created")
        dg_min_dt, dg_max_dt = up_minmax(tc, dg_min_dt, dg_max_dt)

        if u:
            h = url_hash(u)
            buf_url.append((h, u))
            urls_seen += 1
        if t and c and tc:
            s = soft_key(c, t, tc)
            buf_soft.append((s, c, t, str(tc)[:10]))
            soft_seen += 1

        if len(buf_url)>=BATCH:
            cur.executemany("INSERT OR IGNORE INTO urls_dg(url_hash, example_url) VALUES (?,?)", buf_url)
            buf_url.clear()
        if len(buf_soft)>=BATCH:
            cur.executemany("INSERT OR IGNORE INTO soft_dg(soft_hash, ex_company, ex_title, ex_day) VALUES (?,?,?,?)", buf_soft)
            buf_soft.clear()

# flush
if buf_url:
    cur.executemany("INSERT OR IGNORE INTO urls_dg(url_hash, example_url) VALUES (?,?)", buf_url)
if buf_soft:
    cur.executemany("INSERT OR IGNORE INTO soft_dg(soft_hash, ex_company, ex_title, ex_day) VALUES (?,?,?,?)")
con.commit()

dg_url_count = cur.execute("SELECT COUNT(*) FROM urls_dg").fetchone()[0]
dg_soft_count = cur.execute("SELECT COUNT(*) FROM soft_dg").fetchone()[0]

print("Indexed with-dg bundle:")
print(f"  files: {len(dg_files)}")
print(f"  date range (tst_created): {dg_min_dt} → {dg_max_dt}")
print(f"  distinct URL hashes: {dg_url_count:,}  (raw seen: {urls_seen:,})")
print(f"  distinct soft keys:  {dg_soft_count:,}  (raw seen: {soft_seen:,})")

# ---------- Pass 2: scan the OLD file and compare ----------
old_min_dt = None
old_max_dt = None
old_total = 0
old_url_have = 0
old_soft_have = 0

# to show a few examples that matched:
ex_url_hashes = set()
ex_soft_hashes = set()

def batch_membership(table, col, keys):
    # returns set(keys_present)
    if not keys: return set()
    # chunk to stay under SQLite variable limits (~999)
    present = set()
    for i in range(0, len(keys), 800):
        sub = keys[i:i+800]
        qmarks = ",".join(["?"]*len(sub))
        if table=="urls_dg":
            rows = cur.execute(f"SELECT {col} FROM urls_dg WHERE {col} IN ({qmarks})", sub).fetchall()
        else:
            rows = cur.execute(f"SELECT {col} FROM soft_dg WHERE {col} IN ({qmarks})", sub).fetchall()
        present.update(r[0] for r in rows)
    return present

buf_old_url = []
buf_old_soft = []

for rec in iter_records(OLD_FILE):
    old_total += 1
    u = rec.get("url")
    t = rec.get("title")
    c = rec.get("company_name")
    tc = rec.get("tst_created")
    old_min_dt, old_max_dt = up_minmax(tc, old_min_dt, old_max_dt)

    if u:
        buf_old_url.append(url_hash(u))
    if t and c and tc:
        buf_old_soft.append(soft_key(c, t, tc))

    if len(buf_old_url) >= 5000 or len(buf_old_soft) >= 5000:
        pres = batch_membership("urls_dg", "url_hash", buf_old_url)
        old_url_have += len(pres)
        ex_url_hashes |= set(list(pres)[:5])
        buf_old_url.clear()

        pres = batch_membership("soft_dg", "soft_hash", buf_old_soft)
        old_soft_have += len(pres)
        ex_soft_hashes |= set(list(pres)[:5])
        buf_old_soft.clear()

# flush remainder
if buf_old_url:
    pres = batch_membership("urls_dg", "url_hash", buf_old_url)
    old_url_have += len(pres)
    ex_url_hashes |= set(list(pres)[:5])

if buf_old_soft:
    pres = batch_membership("soft_dg", "soft_hash", buf_old_soft)
    old_soft_have += len(pres)
    ex_soft_hashes |= set(list(pres)[:5])

print("\nOld file scan:")
print(f"  file: {os.path.basename(OLD_FILE)}")
print(f"  date range (tst_created): {old_min_dt} → {old_max_dt}")
print(f"  records scanned: {old_total:,}")

pct_url = 100.0*old_url_have/old_total if old_total else 0.0
pct_soft = 100.0*old_soft_have/old_total if old_total else 0.0
print(f"\nOverlap by URL:  {old_url_have:,}/{old_total:,}  ({pct_url:.1f}%)")
print(f"Overlap by soft:  {old_soft_have:,}/{old_total:,}  ({pct_soft:.1f}%)")

# show a few examples
if ex_url_hashes:
    print("\nExamples (URL matches):")
    for h in list(ex_url_hashes)[:5]:
        row = cur.execute("SELECT example_url FROM urls_dg WHERE url_hash=?", (h,)).fetchone()
        if row:
            print("  ", row[0])

if ex_soft_hashes:
    print("\nExamples (soft matches: company | title | day):")
    for h in list(ex_soft_hashes)[:5]:
        row = cur.execute("SELECT ex_company, ex_title, ex_day FROM soft_dg WHERE soft_hash=?", (h,)).fetchone()
        if row:
            print("  ", " | ".join(row))
```

```python
def parse_x28_occupations(x28_str):
    """Parse X28 occupations field into list of clean strings"""
    # Handle PostgreSQL arrays (which come as numpy arrays or lists)
    if hasattr(x28_str, '__iter__') and not isinstance(x28_str, str):
        # It's already an array/list from PostgreSQL
        return [str(x).strip() for x in x28_str if x and str(x).strip()]

    # Handle null/empty values (but only for strings)
    try:
        if not x28_str or pd.isna(x28_str) or str(x28_str).strip() in ('', 'null', 'None'):
            return []
    except ValueError:
        # If we get here, x28_str is likely an array that wasn't caught above
        if hasattr(x28_str, '__len__') and len(x28_str) == 0:
            return []
        # For non-empty arrays, convert to string list
        return [str(x).strip() for x in x28_str if x and str(x).strip()]

    x28_str = str(x28_str).strip()

    # Handle JSON-style arrays: ['123', '456'] or ["123", "456"]
    if x28_str.startswith('[') and x28_str.endswith(']'):
        try:
            # Try JSON parsing first
            parsed = json.loads(x28_str)
            if isinstance(parsed, list):
                return [str(x).strip().strip('\'"') for x in parsed if x]
        except:
            # Fallback: manual parsing for space-separated or comma-separated values
            x28_str = x28_str[1:-1]  # Remove brackets

            # Split by comma first, then by space for each part
            codes = []
            for part in x28_str.split(','):
                part = part.strip().strip('\'"')
                if part:
                    # Handle space-separated codes within this part
                    for code in part.split():
                        code = code.strip().strip('\'"')
                        if code:
                            codes.append(code)

            # If no comma-separated values found, try space-separated
            if not codes:
                for code in x28_str.split():
                    code = code.strip().strip('\'"')
                    if code:
                        codes.append(code)

            return codes

    # Handle comma-separated values
    codes = []
    for code in x28_str.split(','):
        code = code.strip().strip('\'"')
        if code:
            # Also handle space-separated codes within each comma-separated part
            for subcode in code.split():
                subcode = subcode.strip().strip('\'"')
                if subcode:
                    codes.append(subcode)

    return codes
    


df_j = df.head(50)
df_j['x28_codes'] = df_j['x28_occupations'].apply(parse_x28_occupations)
df_jobs_exploded = df_j.explode('x28_codes').reset_index(drop=True)

df_jobs_exploded
```
